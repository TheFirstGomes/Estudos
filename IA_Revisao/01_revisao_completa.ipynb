```python
'''
1. Implemente uma função que receba uma lista de inteiros e retorne:
soma
média
maior valor
menor valor
Tudo em uma única estrutura de retorno.
'''

def calculos(lista: list):
    soma = sum(lista)
    media = sum(lista) / len(lista)
    maior = max(lista)
    menor = min(lista)

    return f'A soma é: {soma}. A média é: {media}. O maior núm: {maior}, O menor núm: {menor}'

lista = [1, 2, 3, 4, 5]
calculos(lista)
```

```python
'''
2. Dada uma string arbitrária, determine se ela é um palíndromo, ignorando:
espaços
capitalização
caracteres não alfanuméricos.
'''

def frase(texto:str):
    palindromo = ''.join(c.lower() for c in texto if c.isalnum())
    return palindromo == palindromo[::-1]

print(frase('Arara'))
print(frase('Pastel'))
```

```python
'''
3. Crie uma função que receba um número inteiro positivo n e retorne os n primeiros números da sequência de Fibonacci.
'''
def fibonacci(num:int):
    if num <= 0:
        return None
    elif num == 1:
        return [0]
    elif num == 2:
        return [0, 1]

    seq = [0, 1]
    for _ in range(2, num):
        seq.append(seq[-1] + seq[-2])
    return seq

fibonacci(4)
```

```python
'''
Dada uma lista de strings, retorne um dicionário onde:
chave → string original
valor → quantidade de vogais presentes.
'''

lista = ['frase', 'amarelo', 'lista']

def conversao(lista:list):
    vogais = {v: 0 for v in 'aeiou'}

    for palavra in lista:
        if palavra in vogais:
            vogais[palavra] += 1
    return vogais


conversao(lista)
```
