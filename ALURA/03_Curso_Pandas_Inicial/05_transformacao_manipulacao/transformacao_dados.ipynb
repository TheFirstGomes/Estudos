# 1. Import dos dados e conhecimento do problema

A precificação inteligente para hospedagem é uma estratégia de estimação de preços de forma automatizada e dinâmica, que considera fatores como oferta e demanda, sazonalidade, eventos locais, características do local, entre outros. Com base nessas informações, um algoritmo pode ajustar os preços para maximizar a receita e a rentabilidade da pessoa proprietária.

```python
import pandas as pd

dados_hospedagem = r'C:\Users\Usuario\Documents\Estudos\FIAP\Estudos\ALURA\03_Curso_Pandas_Inicial\raw\dados_hospedagem.json'
dados_moveis = r'C:\Users\Usuario\Documents\Estudos\FIAP\Estudos\ALURA\03_Curso_Pandas_Inicial\raw\moveis_disponiveis.json'

df_hospedagem = pd.read_json(dados_hospedagem)
df_moveis = pd.read_json(dados_moveis)
```

1. Normalizar a base de dados df_hospedagem

```python
df_hospedagem = pd.json_normalize(df_hospedagem['info_moveis'])
df_hospedagem.head()
```

# 2. Transformação dos dados das listas aninhadas

```python
col_hospedagem = list(df_hospedagem.columns)
col_hospedagem
```

```python
df_hospedagem = df_hospedagem.explode(col_hospedagem[3:])
df_hospedagem
```

Resetar os indexes do dataframe, pois como mostrado acima, pode vir tudo 0 por causa das linhas como o 69 para várias linhas

```python
df_hospedagem.reset_index(inplace=True, drop=True)
df_hospedagem.head(2)
```

```python
df_hospedagem.info()
```

#### Convertendo dados númericos

```python
import numpy as np
```

Forma de coluna a coluna

```python
df_hospedagem['max_hospedes'] = df_hospedagem['max_hospedes'].astype(np.int8)
```

```python
df_hospedagem.info()
```

Forma mais rápida de se fazer para fazer em várias colunas: 

```python
col_hosp_numericas = ['quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']
df_hospedagem[col_hosp_numericas] = df_hospedagem[col_hosp_numericas].astype(np.int8)
```

```python
df_hospedagem['avaliacao_geral'] = df_hospedagem['avaliacao_geral'].astype(np.float16)
```

```python
df_hospedagem.info()
```

Números em strings - tratando a coluna preço que vem -> $100.00

```python
df_hospedagem['preco'] = df_hospedagem['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
df_hospedagem['preco'] = df_hospedagem['preco'].astype(np.float16)
df_hospedagem.info()
```

Números em strings - tratando as colunas taxa_deposito e taxa_limpeza que vem -> $100.00

```python
df_hospedagem[['taxa_deposito', 'taxa_limpeza']] = df_hospedagem[['taxa_deposito', 'taxa_limpeza']].map(lambda x: x.replace('$', '').replace(',', '').strip())
df_hospedagem[['taxa_deposito', 'taxa_limpeza']] = df_hospedagem[['taxa_deposito', 'taxa_limpeza']].astype(np.float16)

df_hospedagem.info()
```

# 3. Dados textuais

```python
df_hospedagem['descricao_local'] = df_hospedagem['descricao_local'].str.lower()
```

```python
df_hospedagem.head(2)
```

```python
df_hospedagem['descricao_local'][3169]
```

```python
df_hospedagem['descricao_local'] = df_hospedagem['descricao_local'].str.replace("[^a-zA-Z0-9\-\']", ' ', regex=True)
```

```python
# Verificar se existe um caractere antes e depois do hífen
df_hospedagem['descricao_local'] = df_hospedagem['descricao_local'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
```

Tokenização de strings

```python
df_hospedagem['descricao_local'] = df_hospedagem['descricao_local'].str.split()
df_hospedagem.head()
```

```python
df_hospedagem['comodidades'] = df_hospedagem['comodidades'].str.replace('\{|}|\"', '', regex=True)
```

```python
df_hospedagem['comodidades'] = df_hospedagem['comodidades'].str.split(',')
df_hospedagem.head()
```

```python
df_hospedagem['descricao_vizinhanca'] = df_hospedagem['descricao_vizinhanca'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
df_hospedagem['descricao_vizinhanca'] = df_hospedagem['descricao_vizinhanca'].str.lower()
```

```python
df_hospedagem['descricao_vizinhanca'] = df_hospedagem['descricao_vizinhanca'].str.split()
df_hospedagem
```
