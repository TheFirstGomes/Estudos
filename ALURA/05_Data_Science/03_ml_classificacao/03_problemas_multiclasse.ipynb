# Realizando o treinamento

## Preparando os dados

```python
dados.columns
```

```python
from sklearn.preprocessing import OneHotEncoder

colunas_categoricas = ['Migração', 'Sexo','Estrangeiro', 'Necessidades educacionais especiais', 'Devedor',
                     'Taxas de matrícula em dia', 'Bolsista', 'Período','Estado civil', 'Curso', 'Qualificação prévia']

oht = OneHotEncoder(drop='if_binary')

df_categorico = dados[colunas_categoricas]
df_oht = pd.DataFrame(oht.fit_transform(df_categorico).toarray(), columns=oht.get_feature_names_out(colunas_categoricas))

df_final = pd.concat([dados.drop(columns=colunas_categoricas, axis=1), df_oht], axis=1)
```

```python
df_final.sample(5)
```

## Separando os dados

```python
from sklearn.model_selection import train_test_split

X = df_final.drop(columns='Target')
y = df_final['Target']

X, X_teste, y, y_teste = train_test_split(X, y, test_size=0.15, stratify=y, random_state=0)
X_treino, X_val, y_treino, y_val = train_test_split(X, y, stratify=y, random_state=0)
```

## Treinado Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(random_state=0, max_depth=10)
model_rf.fit(X_treino, y_treino)

y_pred = model_rf.predict(X_val)
print(f'Acurácia de Treino: {model_rf.score(X_treino, y_treino)}')
print(f'Acurácia de Validação: {model_rf.score(X_val, y_val)}')
```

## Desafio da Aula

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(random_state=0, solver='lbfgs', max_iter=2500).fit(X_treino, y_treino)
y_pred = model_lr.predict(X_val)
print(f'Acurácia de Treino: {model_lr.score(X_treino, y_treino)}')
print(f'Acurácia de Validação: {model_lr.score(X_val, y_val)}')
```

# Avaliando o Modelo

## Analisando a matriz de confusão

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
```

```python
matriz_confusao = confusion_matrix(y_true=y_val, y_pred=y_pred)
visualizacao = ConfusionMatrixDisplay(confusion_matrix=matriz_confusao, display_labels=model_rf.classes_)
visualizacao.plot();
```

```python
ConfusionMatrixDisplay.from_predictions(y_true=y_val, y_pred=y_pred, normalize='true', cmap='Blues');
```

## Analisando o relatório de métricas

```python
from sklearn.metrics import classification_report

report = classification_report(y_true=y_val, y_pred=y_pred)
print(f'Relatório de Classificação: \n {report}')
```

## Superamostragem -> Oversample

```python
y_treino.value_counts()
```

```python
from imblearn.over_sampling import SMOTE

oversample = SMOTE(random_state=0)
X_balanceado, y_balanceado = oversample.fit_resample(X_treino, y_treino)

y_balanceado.value_counts()
```

```python
dtc = RandomForestClassifier(random_state=0, max_depth=10)
dtc.fit(X_balanceado, y_balanceado)
y_pred = dtc.predict(X_val)
```

## Analisando métricas após o balanceamento

```python
print(classification_report(y_val, y_pred))
ConfusionMatrixDisplay.from_predictions(y_val, y_pred, normalize='true', cmap='Blues');
```

## Desafio da Aula

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(random_state=0, solver='lbfgs', max_iter=2500).fit(X_balanceado, y_balanceado)
y_pred = model_lr.predict(X_val)
print(f'Acurácia de dados Balanceados: {model_lr.score(X_balanceado, y_balanceado)}')
print(f'Acurácia de dados Validação: {model_lr.score(X_val, y_val)}')

ConfusionMatrixDisplay.from_estimator(model_lr, X_val, y_val, normalize = 'true', cmap = 'Blues');

report = classification_report(y_val, y_pred)
print("\nRelatório de Classificação:\n", report)
```

# Validando o modelo

## Pipeline para validação de dados desbalanceados

```python
from imblearn.pipeline import Pipeline as imbpipeline

rf = RandomForestClassifier(max_depth=10)
pipeline = imbpipeline([('oversample', SMOTE()), ('floresta', rf)])
```

```python
from sklearn.model_selection import cross_validate, StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
cv_resultados = cross_validate(pipeline, X, y, cv=skf, scoring='recall_weighted')
cv_resultados['test_score']
```

## Obtendo o intervalo de confiança

```python
media = cv_resultados['test_score'].mean()
desvio_padrao = cv_resultados['test_score'].std()

print(f'Limite de confiança: [{media-2*desvio_padrao}, {min(media+2*desvio_padrao, 1.0)}]')
```

# Testando o modelo

```python
oversample = SMOTE(random_state=0)
X_balanceado, y_balanceado = oversample.fit_resample(X, y)

rf = RandomForestClassifier(random_state=0, max_depth=10)
rf.fit(X=X_balanceado, y=y_balanceado)
y_pred = rf.predict(X_teste)

print(classification_report(y_true=y_teste, y_pred=y_pred))
ConfusionMatrixDisplay.from_predictions(y_true=y_teste, y_pred=y_pred, normalize='true', cmap='Blues');
```
