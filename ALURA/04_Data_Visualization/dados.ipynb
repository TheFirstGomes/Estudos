# 1° Carregando e visualizando os dados

```python
import pandas as pd
```

```python
url = r'C:\Users\Usuario\Documents\Estudos\FIAP\Estudos\ALURA\04_Data_Visualization\raw\imigrantes_canada.csv'
df = pd.read_csv(url, encoding='utf-8')
df
```

```python
df.info()
```

### Mudar o index para o nome dos paises e pegar somente os anos

```python
df.set_index('País', inplace=True)
```

```python
brasil = df.loc['Brasil', '1980':'2013']
```

```python
brasil_dict = {'ano':brasil.index.tolist(), 'imigrantes':brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil.head()
```

# 2° Trabalhando com o primeiro gráfico

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('N° de imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
# plt.yticks([500, 1000, 1500, 2000, 2500, 3000]);
plt.show()
```

# 3° Criando figuras

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
ax.set_title('Imigração do Brasil para o Canadá\n1980 à 2013')
ax.set_xlabel('Ano')
ax.set_ylabel('N° Imigrantes')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(15,5))

axs[0].plot(dados_brasil['ano'], dados_brasil['imigrantes'])
axs[0].set_title('Imigração do Brasil para o Canadá\n1980 à 2013')
axs[0].set_xlabel('Ano')
axs[0].set_ylabel('N° imigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(dados_brasil['imigrantes'])
axs[1].set_title('Boxplot da imigração do Brasil para o Canadá\n1980 à 2013')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('N° imigrantes')
axs[1].grid()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 9))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Imigração dos quatro maiores países da América do Sul para o Canadá de 1980 à 2013')

axs[0,0].plot(df.loc['Brasil', '1980':'2013'])
axs[0,0].set_title('Brasil')

axs[0,1].plot(df.loc['Colômbia', '1980':'2013'])
axs[0,1].set_title('Colômbia')

axs[1,0].plot(df.loc['Argentina', '1980':'2013'])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Peru', '1980':'2013'])
axs[1,1].set_title('Peru')

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))

for ax in axs.flat:
    ax.set_xlabel('Ano')
    ax.set_ylabel('N° Imigrantes')

ymin = 0 
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)

plt.show()
```

# 4° Customização do Matplotlib

```python
with plt.style.context('fivethirtyeight'):
    fig, ax = plt.subplots(figsize=(8,4))
    ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3, color='g') #marker='o')
    ax.set_title('Imigração do Brasil para o Canadá:\n1980 à 2013', fontsize=18, loc='left')
    ax.set_xlabel('Ano', fontsize=14)
    ax.set_ylabel('N° Imigrantes', fontsize=14)
    ax.xaxis.set_tick_params(labelsize=12)
    ax.yaxis.set_tick_params(labelsize=12)
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.show()
```

```python
df.head(2)
```

```python
america_sul = df.query('Região == "América do Sul"').sort_values('Total', ascending=True)
america_sul
```

```python
#cores = ['royalblue', 'orange', 'forestgreen', 'orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal', 'tomato']
cores = []

for pais in america_sul.index:
    if pais == 'Brasil': 
        cores.append('green')
    else: 
        cores.append('silver')

fig, ax = plt.subplots(figsize=(12, 5))
ax.barh(america_sul.index, america_sul['Total'], color=cores)
ax.set_title('América do Sul: Brasil foi o 4° país com mais imigrantes\npara o Canadá no período de 1980 à 2013', loc='left', fontsize=15)
ax.set_xlabel('N° de Imigrantes', fontsize=10)
ax.set_ylabel('')
ax.xaxis.set_tick_params(labelsize=9)
ax.yaxis.set_tick_params(labelsize=9)

for i, v in enumerate(america_sul['Total']):
    ax.text(v + 20, i, str(v), color='black', fontsize=10, ha='left', va='center')

ax.set_frame_on(False)
ax.get_xaxis().set_visible(False)
ax.tick_params(axis='both', which='both', length=0)

fig.savefig('imigracao_brasil_canada.png', transparent=False, dpi=300, bbox_inches='tight')
plt.show()
```

# 5° Trabalhando com o SEABORN

```python
import seaborn as sns
```

```python
sns.set_theme()
```

```python
top_10_imigracoes = df.sort_values('Total', ascending=False).head(10)
top_10_imigracoes
```

```python
sns.barplot(data=top_10_imigracoes, x=top_10_imigracoes.index, y='Total');
```

```python
sns.barplot(data=top_10_imigracoes, x='Total', y=top_10_imigracoes.index, orient='h');
```

```python
ax = sns.barplot(data=top_10_imigracoes, x='Total', y=top_10_imigracoes.index, orient='h')
ax.set(title='Países com maior imigração para o Canadá\n1980 - 2013', 
       xlabel='N° de Imigrantes', 
       ylabel='')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax = sns.barplot(data=top_10_imigracoes, x='Total', y=top_10_imigracoes.index, orient='h')

ax.set_title('Países com maior imigração para o Canadá\n1980 - 2013', loc='left', fontsize=14)
ax.set_xlabel('N° de Imigrantes', fontsize=10)
ax.set_ylabel('')
plt.show()
```

```python
def gerar_grafico_paleta(palette): 
    fig, ax = plt.subplots(figsize=(8,4))
    ax = sns.barplot(data=top_10_imigracoes, x='Total', y=top_10_imigracoes.index, orient='h', palette=palette, legend=False)
    ax.set_title('Países com maior imigração para o Canadá\n1980 - 2013', loc='left', fontsize=14)
    ax.set_xlabel('N° de Imigrantes', fontsize=10)
    ax.set_ylabel('')
    plt.show();
```

# 6° Gráfico interativo com PLOTLY

```python
import plotly.express as px
```

```python
fig = px.line(dados_brasil, x='ano', y='imigrantes', title='Imigração do Brasil para o Canadá no período de 1980 à 2013')
fig.update_traces(line_color='green')
fig.update_layout(width=1000, height=500, 
                  xaxis={'tickangle':-45}, 
                  font_family='Arial', font_size=14, font_color='grey', 
                  title_font_color='black', title_font_size=22, 
                  xaxis_title='Ano', yaxis_title='N° Imigrantes')
fig.show()
```
