# Exemplo de Chatbot

```python
# !pip install nltk --quiet
import nltk 
from nltk.chat.util import Chat
```

```python
reflections_personalizados_somente_nas_respostas_do_chat = {
    'eu':'você', 
    'meu': 'seu', 
    'minha': 'sua', 
    'meus': 'seus',
    'minhas': 'suas',
    'seu': 'meu',
    'sua': 'minha',
    'seus': 'meus',
    'suas': 'minhas',
    'vou': 'vai', 
    'era': 'fui'
}

pares_de_conversacao = [
    [
        r'meu nome é (.*)', # Captura o nome do usuário
        ['Olá %1, como posso ajudar você?']
    ], 
    [
        r'eu sou (.*)', # Captura a informação do usuário
        ['Por que você é %1? Me conte mais sobre isso.']
    ], 
    [
        r'meu (.*) é (.*)', # Captura algo que o usuário possui
        ['Por que meu %1 é %2? Me conte mais sobre isso.']
    ], 
    [
        r'como você está?', # Pergunta sobre o estado do chatbot
        ['Estou bem, obrigado por perguntar! E você?']
    ], 
    [
        r'(.*) você gosta de (.*)', # Pergunta sobre os gostos do chatbot
        ['Não sei se gosto de %2, mas estou sempre aberto a aprender sobre novos interesses! E você?']
    ], 
    [
        r'sair', # Comando para sair da conversa
        ['Tchau! Foi um prazer conversar com você. Até a próxima!']
    ], 
    [
        r'(.*)', 
        ['Desculpe, não entendi isso. Pode reformular a pergunta?']
    ]
]
```

```python
def aplicar_reflections(texto, reflections): 
    palavras = texto.lower().split()
    resposta_transformada = [reflections.get(palavra, palavra) for palavra in palavras]
    return ' '.join(resposta_transformada)

# Criar o chatbot
chatbot = Chat(pares_de_conversacao, reflections_personalizados_somente_nas_respostas_do_chat)
```

```python
print('Chatbot: Olá! Sou um chatbot. Diga ´sair´ para encerrar a conversa.')
while True: 
    entrada = input('Você: ')
    if entrada.lower() == 'sair':
        print('Chatbot: Tchau! Foi um prazer conversar com você. Até a próxima!')
        break
    resposta = chatbot.respond(entrada)
    if resposta: 
        resposta_transformada = aplicar_reflections(resposta, reflections_personalizados_somente_nas_respostas_do_chat)
        print(f'Chatbot: {resposta_transformada}')
    else: 
        print('Chatbot: Desculpe, não entendi isso')
```

# Exemplo de Agente

```python
class Ambiente: 
    def __init__(self): 
        self._casas = [
            ['sujo', 'limpo'], 
            ['sujo', 'sujo']
        ]
    
    def exibir_ambiente(self): 
        print('Estado atual do ambiente:')
        for linha in self._casas: 
            print(linha) 
    
class Agente: 
    def __init__(self, ambiente): 
        self._ambiente = ambiente
        self._posicao = [0, 0]

    def perceber(self): 
        x, y = self._posicao
        return self._ambiente._casas[x][y]
    
    def decidir_acao(self, estado): 
        if estado == 'sujo': 
            return 'limpar'
        return 'mover'
    
    def agir(self, acao): 
        # Executa a ação decididad
        if acao == 'limpar': 
            x, y = self._posicao 
            self._ambiente._casas[x][y] = 'limpo'
            print(f'Limpando a posição {self._posicao}')
        elif acao == 'mover':
            self.mover()
    
    def mover(self): 
        # Move o agente para a próxima posição (simplesmente alternando entre as casas)
        x, y = self._posicao
        if y < len(self._ambiente._casas[0]) - 1: # Move para a direita, se possível
            self._posicao = [x, y + 1] 
        elif x < len(self._ambiente._casas) - 1: # Move para baixo, se possível
            self._posicao = [x + 1, 0]
        else: 
            print(f'Fim do ambiente. Nada mais para fazer')
            self._posicao = None 
        print(f'Movendo para a posição {self._posicao}')

    def executar(self):
        while self._posicao is not None: 
            estado = self.perceber()
            acao = self.decidir_acao(estado)
            self.agir(acao)
            self._ambiente.exibir_ambiente()
```

```python
# Criação do agente e ambiente
ambiente = Ambiente()
agente = Agente(ambiente)
ambiente.exibir_ambiente()
agente.executar()
```
