# Import das bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Trazer os dados

```python
uri = r'C:\Users\Usuario\Documents\Estudos\FIAP\Fase3\Cap_01_Etapas_Maquina_Agricola\raw\dados_irrigacao.csv'
df = pd.read_csv(uri, parse_dates=['timestamp'])

df.sample(5)
```

# Criação do Gráfico

### Distribuição da Umidade

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(8,5))
sns.histplot(df['umidade'], bins=30, color='skyblue', kde=True)
plt.title('Distribuição da Umidade do Solo')
plt.xlabel('Umidade (%)')
plt.ylabel('Frequência')
plt.tight_layout()
plt.show()
```

### Série temporal da Umidade

```python
plt.figure(figsize=(10,5))
sns.lineplot(x="timestamp", y="umidade", data=df.sample(1000))  # amostra para performance
plt.title("Evolução da Umidade ao Longo do Tempo")
plt.xlabel("Data")
plt.ylabel("Umidade (%)")
plt.tight_layout()
plt.show()
```

### Relação entre ph e status de irrigação 

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="estado_bomba", y="ph", data=df, palette="viridis")
plt.title("pH x Estado da Irrigação")
plt.xlabel("Bomba (0=Desligada / 1=Ligada)")
plt.ylabel("pH")
plt.tight_layout()
plt.show();
```
